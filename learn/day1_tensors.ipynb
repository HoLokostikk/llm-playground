```
x = torch.tensor(3.0, requires_grad = True)
y = x ** 2
y.backward()
print(x.grad)
```

```
x = torch.tensor(3.0, requires_grad = True)
y = 3 * x ** 2 + 2 * x
y.backward()
print(x.grad)
```

```
x = torch.tensor(0.0, requires_grad=True)
lr = 0.01

for step in range(100):
    loss = (x - 5)**2

    loss.backward()

    with torch.no_grad():
        x -= lr * x.grad
    x.grad.zero_()
    if step % 10 == 0:
        print(f"step : {step}, loss : {loss.item():.4f}, x = {x.item():.4f}")
```
