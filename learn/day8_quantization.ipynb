```
import torch
torch.manual_seed(0)
w = torch.randn(8) * 0.05

scale = w.abs().max() / 127

q = torch.round(w / scale).to(torch.int8)
w_back = q.float() * scale

print("оригінал:   ", w)
print("цілі int8:  ", q)
print("відновлене: ", w_back)
print("макс. помилка:", (w - w_back).abs().max().item())
```

```
scale4 = w.abs().max() / 7
q4 = torch.round(w / scale4).clamp(-7, 7)
w_back4 = q4 * scale4
print("цілі int4:  ", q4)
print("макс. помилка int4:", (w - w_back4).abs().max().item())
```

```
w_outlier = w.clone()
w_outlier[0] = 1.0                 # одна вага в 20 разів більша за інші

scale = w_outlier.abs().max() / 7
w_back = torch.round(w_outlier / scale).clamp(-7, 7) * scale
print("відновлене:", w_back)
```
